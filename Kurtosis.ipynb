```python
import pandas as pd
import matplotlib.pyplot as plt
data=pd.read_csv("Skincare_Brands_Sales_Dataset.csv")
data
```

```python
sales=data["Sales"]
kurtosis_value=sales.kurtosis()
print("Kurtosis Value:",kurtosis_value)

if kurtosis_value>0:
    print("Leptokurtosis.")
elif kurtosis_value<0:
    print("Platykurtosis.")
else:
    print("Mesokurtosis.")
```

```python
plt.figure(figsize=(10,10))
plt.hist(sales,bins=13)
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Histogram of Sales")
plt.show()
```

```python
discount=data["Discount"]
kurtosis_value=discount.kurtosis()
print("Kurtosis Value:",kurtosis_value)

if kurtosis_value>0:
    print("Leptokurtosis.")
elif kurtosis_value<0:
    print("Platykurtosis.")
else:
    print("Mesokurtosis.")
```

```python
plt.figure(figsize=(10,10))
plt.hist(discount,bins=6)
plt.xlabel("Discount")
plt.ylabel("Frequency")
plt.title("Histogram of Discount")
plt.show()
```
