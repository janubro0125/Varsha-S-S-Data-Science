```python
import matplotlib.pyplot as plt
import numpy as np
days=np.array(["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"])
sales=np.array([20000,10000,25000,30000,45000,50000,60000])
plt.pie(sales,
    labels=days,
    colors=["lavender","green","red","Blue","Yellow","Pink","Grey"]
    )
plt.title("Sales Analysis")
plt.show()
```
