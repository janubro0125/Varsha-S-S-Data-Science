```python
import numpy as np
import pandas as pd
a=pd.read_csv("Skincare_Brands_Sales_Dataset.csv")
a
```

```python
data=a["Sales"]
data
```

```python
from scipy.stats import skew
answer=skew(data)
if answer<0:
  print("Negative Skewness")
elif answer>0:
  print("Positive Skewness")
else:
  print("Symmetric Skewness")
```

```python
import matplotlib.pyplot as plt
plt.hist(data)
plt.title("Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```
